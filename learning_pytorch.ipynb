#PyTorch Lesson 4: Autograd and Gradients

###Welcome to Lesson 4! You're making good progress. So far, you've learned how to create tensors, select elements, and perform calculations. Now we're going to learn one of the most important concepts in deep learning: how PyTorch calculates gradients so a model can learn from its mistakes.
1. What is a gradient?
Imagine you're training a machine learning model to predict internet bandwidth requirements. Its predictions are initially inaccurate, so it needs to adjust its internal parameters to reduce its errors.

A gradient tells us how a quantity changes when an input or parameter changes. During training, gradients help the model determine how to adjust its parameters to reduce the loss.

##This tells PyTorch to track operations involving that tensor so it can calculate gradients later.


```python
import torch

x = torch.tensor(3.0, requires_grad=True)

y = x ** 2

print("x:", x)
print("y:", y)
```

```python
import torch

x = torch.tensor(5.0, requires_grad=True)
y = x ** 2

y.backward()

print("Output:", y.item())
print("Gradient:", x.grad.item())
```

```python
import torch
import torch.nn as nn

class SimpleModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer = nn.Linear(1, 1)

    def forward(self, x):
        return self.layer(x)

model = SimpleModel()

x = torch.tensor([[5.0]])
prediction = model(x)

print("Prediction:", prediction)
```

# PyTorch Lesson 6: Training a Neural Network
Welcome to Lesson 6! You've learned how to create a model and make predictions. Now we'll learn how to train it so its predictions become more accurate.

This is a major step toward building real ML/AI projects, including your bandwidth optimization and classification projects.

Today's goal: Understand the four key parts of training: predictions, loss, gradients, and optimizer updates.
1. How does a neural network learn?
Imagine you want to build a model that predicts internet bandwidth demand based on the number of connected users.

Suppose the actual bandwidth demand is 11 Mbps, but your model predicts 7 Mbps. The prediction is wrong by 4 Mbps.

A training process repeatedly does four things:

```python
import torch
import torch.nn as nn

# Training data
X = torch.tensor([[1.0], [2.0], [3.0], [4.0]])
y = torch.tensor([[2.0], [4.0], [6.0], [8.0]])

# Build the model
model = nn.Linear(1, 1)

# Choose the loss function and optimizer
loss_fn = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)

# Train the model
for epoch in range(500):
    # 1. Make predictions
    predictions = model(X)

    # 2. Calculate loss
    loss = loss_fn(predictions, y)

    # 3. Clear previous gradients
    optimizer.zero_grad()

    # 4. Calculate gradients
    loss.backward()

    # 5. Update the model parameters
    optimizer.step()

print("Final loss:", loss.item())
print("Prediction for x=3:", model(torch.tensor([[3.0]])).item())
```
